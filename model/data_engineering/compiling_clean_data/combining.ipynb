```python
import pandas as pd
```

```python
incident_data = pd.read_csv("../../clean_data/florida_incident_data_w_categories.csv")
valid_incident_data = pd.read_csv("../../clean_data/valid_incident_weather_info.csv")
invalid_incident_data = pd.read_csv("../../clean_data/valid_incident_weather_info.csv")
```

```python
valid_incident_info = incident_data[["ID", "incident_type"]]
```

```python
valid_incident_data
```

## Forming dataset

```python
complete_incident_data = pd.merge(left=valid_incident_info, right=valid_incident_data, left_on="ID", right_on="associated_incident_id")
```

```python
complete_incident_data.drop(columns=["date", "ID", "associated_incident_id"], inplace=True)
```

```python
complete_incident_data
```

```python
complete_nonincident_data = invalid_incident_data
```

```python
complete_nonincident_data.drop(columns=["associated_incident_id", "date"], inplace=True)
```

```python
complete_nonincident_data["incident_type"] = "Not Incident"
```

```python
total_data = pd.concat([complete_nonincident_data, complete_incident_data])
```

```python
total_data.dropna(inplace=True)
```

```python
total_data["is_incident"] = total_data["incident_type"] != "Not Incident"
```

```python
total_data.to_csv("clean_dataset_v1.csv", index=False)
```
