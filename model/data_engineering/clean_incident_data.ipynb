```python
# Direct environmental hazard filter (guaranteed outdoor incidents):
is_outdoor_hazard = (
    data['NatureTitle'].str.contains('Heat|sun', case=False, na=False) |
    data['EventTitle'].str.contains('lightning|weather|animal|insect|bee|wasp|snake', case=False, na=False) |
    data['sourceTitle'].str.contains('heat|atmospheric|tree|animal|insect|sun', case=False, na=False)
)

outdoor_hazard_incidents = data[is_outdoor_hazard]
```
