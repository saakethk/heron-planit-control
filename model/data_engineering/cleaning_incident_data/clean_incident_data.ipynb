## Attempting Filtering with System One Decision Models

```python
from thisthat import TypedDecider, Question

decider = TypedDecider.from_pretrained("flock-io/this-that-model-1.2")
```

```python
options = [
  "Lightning",
  "Flood",
  "Extreme Heat",
  "Wildfire",
  "Soil Saturation",
  "Wildlife (Animals, Insects, ect.)",
  "Other (None of the above)"
]

question = Question("Which of these sources likely caused the incident?", options)
```

```python
narrative = data.iloc[0]["FinalNarrative"]
```

```python
def classify(narrative) -> str:
  # classifies category of incident based on narrative

  answer = decider.decide(
      narrative,
      question
  )

  CONFIDENCE_THRESHOLD = 0.8
  if answer.confidence < CONFIDENCE_THRESHOLD:
    return options[-1]

  return options[answer.index]
```

```python
def classify_incident(row):
  # classified incident into what it was likely caused by

  narrative = row["FinalNarrative"]
  return {"incident_type": classify(narrative)}
```

```python
classify(narrative)
```

```python
classify_incident(data.iloc[0])
```

```python
sample = data
```

```python
from tqdm.auto import tqdm

# 1. Register tqdm with pandas
tqdm.pandas(desc="Processing rows")
```

```python
incident_classification = sample.progress_apply(classify_incident, axis=1, result_type="expand")
```

```python
sample_w_ic = sample.join(incident_classification)
```

```python
sample_w_ic.to_csv("checkpoint_incident_classification.csv")
```
